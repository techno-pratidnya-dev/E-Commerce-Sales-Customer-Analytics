```python

# E-commerce Sales & Customer Analytics
# Step 1: Import Required Libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
print("E-commerce Analytics Project Started!")
```

```python
customers = pd.read_csv('/content/customers.csv')
orders = pd.read_csv('/content/orders.csv')
order_items = pd.read_csv('/content/order_items.csv')
products = pd.read_csv('/content/products.csv')

print("All CSV files loaded successfully!")
```

```python

# Step 3: Check Dataset Dimensions

print("Customers Shape:", customers.shape)
print("Orders Shape:", orders.shape)
print("Order Items Shape:", order_items.shape)
print("Products Shape:", products.shape)
```

```python

# Step 4.1: Display First 5 Rows

print("CUSTOMERS DATA")
display(customers.head())

print("ORDERS DATA")
display(orders.head())

print("ORDER ITEMS DATA")
display(order_items.head())

print("PRODUCTS DATA")
display(products.head())
```

```python

# Step 4.2: Dataset Information

print("CUSTOMERS INFO")
customers.info()

print("\nORDERS INFO")
orders.info()

print("\nORDER ITEMS INFO")
order_items.info()

print("\nPRODUCTS INFO")
products.info()
```

```python

# Step 4.3: Statistical Summary

print("CUSTOMERS SUMMARY")
display(customers.describe(include='all'))

print("ORDERS SUMMARY")
display(orders.describe(include='all'))

print("ORDER ITEMS SUMMARY")
display(order_items.describe())

print("PRODUCTS SUMMARY")
display(products.describe(include='all'))

```

```python

# Step 5.1: Missing Values Check

print("Customers Missing Values:")
display(customers.isnull().sum())

print("Orders Missing Values:")
display(orders.isnull().sum())

print("Order Items Missing Values:")
display(order_items.isnull().sum())

print("Products Missing Values:")
display(products.isnull().sum())
```

```python

# Step 5.2: Duplicate Records Check

print("Customers Duplicates:", customers.duplicated().sum())
print("Orders Duplicates:", orders.duplicated().sum())
print("Order Items Duplicates:", order_items.duplicated().sum())
print("Products Duplicates:", products.duplicated().sum())
```

```python

# Step 5.3: Check Data Types

print("Customers Data Types:")
print(customers.dtypes)

print("\nOrders Data Types:")
print(orders.dtypes)

print("\nOrder Items Data Types:")
print(order_items.dtypes)

print("\nProducts Data Types:")
print(products.dtypes)
```

```python

# Step 6.2: Validate Unique IDs

print("Unique Customer IDs:", customers['customer_id'].nunique())
print("Unique Order IDs:", orders['order_id'].nunique())
print("Unique Order Item IDs:", order_items['order_item_id'].nunique())
print("Unique Product IDs:", products['product_id'].nunique())
```

```python

# Step 6.3: Convert Date Columns

customers['signup_date'] = pd.to_datetime(customers['signup_date'])
orders['order_date'] = pd.to_datetime(orders['order_date'])

print("Date columns converted successfully!")
```

```python
# Step 7: Feature Engineering

orders['year'] = orders['order_date'].dt.year
orders['month'] = orders['order_date'].dt.month
orders['month_name'] = orders['order_date'].dt.month_name()
orders['day'] = orders['order_date'].dt.day

display(orders.head())
```

```python
# Step 8: Sales Summary

print("Total Sales:", order_items['sales_amount'].sum())
print("Total Profit:", order_items['profit'].sum())
print("Total Quantity Sold:", order_items['quantity'].sum())
```

```python

# Step 9: Validate Relationships Between Datasets

invalid_customers = ~orders['customer_id'].isin(
    customers['customer_id']
)

invalid_orders = ~order_items['order_id'].isin(
    orders['order_id']
)

invalid_products = ~order_items['product_id'].isin(
    products['product_id']
)

print("Invalid Customer References:", invalid_customers.sum())
print("Invalid Order References:", invalid_orders.sum())
print("Invalid Product References:", invalid_products.sum())
```

```python
# Step 10 : Data Cleaning Summary

cleaning_summary = pd.DataFrame({
    'Check': [
        'Missing Values in Order Items',
        'Duplicate Rows in Order Items',
        'Invalid Customer References',
        'Invalid Order References',
        'Invalid Product References',
        'Negative Profit Records'
    ],
    'Result': [
        order_items.isnull().sum().sum(),
        order_items.duplicated().sum(),
        invalid_customers.sum(),
        invalid_orders.sum(),
        invalid_products.sum(),
        (order_items['profit'] < 0).sum()
    ]
})

print(cleaning_summary)
```

Step 11: Category-wise Sales Analysis

Business Question:
Which product category generates the highest sales?

```python
category_data = pd.merge(order_items, products, on='product_id')

sales = category_data.groupby('category')['sales_amount'].sum()

print(sales.sort_values(ascending=False))
```

**Business Insight:**

Electronics is the highest-selling category, generating total sales of ₹54,69,410.64. Furniture ranks second with sales of ₹50,51,495.29. Stationery has the lowest sales at ₹2,82,735.91.

**Recommendation:** The business can investigate why Electronics performs well and explore opportunities to improve Stationery sales.

Step 12: Category-wise Profit Analysis

Business Question: Which product category generates the highest profit?

```python
# Category-wise Profit Analysis

category_data = pd.merge(order_items, products, on='product_id')

profit = category_data.groupby('category')['profit'].sum()

print(profit.sort_values(ascending=False))
```

**Business Insight:**

Furniture generates the highest total profit of ₹16,90,475.50, followed by Electronics with ₹16,31,363.06. Stationery generates the lowest total profit of ₹88,191.61.

**Recommendation:** The business should analyze the reasons behind Furniture's strong profitability and identify opportunities to improve Stationery's performance.

Step 13: Monthly Sales Analysis

Business Question: Which month has the highest sales?

```python

# Monthly Sales Analysis

monthly_data = pd.merge(order_items, orders, on='order_id')

monthly_sales = monthly_data.groupby('month_name')['sales_amount'].sum()

print(monthly_sales.sort_values(ascending=False))
```

**Business Insight:**

August recorded the highest sales of ₹23,55,987.30, while November recorded the lowest sales of ₹10,40,407.23.

**Recommendation:** The business should investigate why sales were lower in November and explore ways to improve sales during that month.

### Sales Analysis by Order Status

**Business Question:** How much sales value is associated with each order status?

**Objective:** To understand how sales value is distributed across delivered, shipped, returned, cancelled, and processing orders.

```python
# Sales by Order Status

sales_status = pd.merge(order_items,orders[['order_id', 'order_status']],on='order_id')

status_sales = sales_status.groupby('order_status')['sales_amount'].sum()

print(status_sales.sort_values(ascending=False))
```

### Business Insight

Delivered orders account for the highest sales value of ₹1,12,20,687.62. Returned orders account for ₹28,94,016.80, while cancelled orders account for ₹27,79,567.15.

### Recommendation

The business should investigate the reasons for returned and cancelled orders and review its order fulfilment process to identify opportunities for improvement.

**Note:** These figures represent sales values associated with each order status, not confirmed net revenue or actual financial losses.

Q8 — Customer Segment-wise Sales

Business Question: Which customer segment generates the highest sales?

```python
# Q8: Customer Segment-wise Sales

customer_orders = pd.merge(orders, customers, on='customer_id')

customer_sales = pd.merge(customer_orders, order_items, on='order_id')

segment_sales = customer_sales.groupby('customer_segment')['sales_amount'].sum()

print(segment_sales.sort_values(ascending=False))
```

**Business Insight:**

The Regular customer segment generates the highest sales of ₹80,50,600.93, followed by the New segment with ₹73,27,124.41. The Premium segment generates sales of ₹72,53,252.80.

**Recommendation:** The business should understand the purchasing behaviour of Regular customers and develop strategies to improve sales from New and Premium customers.

Q9 — City-wise Sales Analysis

Business Question: Which city generates the highest sales?

```python
# Q9: City-wise Sales

customer_orders = pd.merge(orders, customers, on='customer_id')

customer_sales = pd.merge(customer_orders, order_items, on='order_id')

city_sales = customer_sales.groupby('city')['sales_amount'].sum()

print(city_sales.sort_values(ascending=False))
```

**Business Insight:**

Mumbai generates the highest sales of ₹19,67,549.86, followed by Thane with ₹19,01,410.46. Solapur has the lowest sales among the listed cities, at ₹12,37,614.33.

**Recommendation:** The business should analyze customer demand in Mumbai and Thane and explore opportunities to improve sales in Solapur.

# EDA Summary — E-commerce Sales & Customer Analytics

## 1. Key Findings

**Sales Analysis**
- Electronics generated the highest category-wise sales of ₹54,69,410.64.
- August recorded the highest monthly sales of ₹23,55,987.30.
- November recorded the lowest monthly sales of ₹10,40,407.23.
- Mumbai generated the highest city-wise sales of ₹19,67,549.86.

**Profit Analysis**
- Furniture generated the highest category-wise profit of ₹16,90,475.50.
- Product P030 ranked first in both sales (₹20,74,302.60) and profit (₹7,94,700.80).
- The dataset contains 32 negative-profit records that require further investigation.

**Customer Analysis**
- The Regular customer segment generated the highest sales of ₹80,50,600.93.
- Credit Card was the most frequently used payment method, with 1,033 orders.

**Order Analysis**
- Delivered was the most common order status, with 2,493 orders.
- There were 628 returned orders and 627 cancelled orders.

## 2. Business Problems Identified

- Monthly sales vary, with November showing the lowest sales.
- Stationery has the lowest category-wise sales and profit.
- Some order items have negative profit.
- Returns and cancellations may indicate opportunities to improve the customer experience.
- Sales performance varies across cities and customer segments.

## 3. Recommendations to Improve Sales

- Analyze low-sales months and plan targeted promotions and seasonal campaigns.
- Improve product visibility and marketing for low-performing categories such as Stationery.
- Use customer purchase patterns to recommend relevant products.
- Encourage repeat purchases through loyalty offers and personalized campaigns.
- Analyze city-wise demand and focus marketing efforts on areas with growth opportunities.

## 4. Recommendations to Improve Profit

- Investigate negative-profit records by checking product costs, selling prices, and discounts.
- Review discount strategies to ensure promotions remain profitable.
- Focus on products and categories with strong profit performance.
- Evaluate product costs and pricing before making price changes.
- Monitor sales and profit together rather than judging performance by sales alone.

## 5. Recommendations to Reduce Returns and Cancellations

- Investigate the reasons behind returned and cancelled orders.
- Review product descriptions, product quality, delivery issues, and customer complaints.
- Improve order tracking and communication with customers.
- Monitor return and cancellation rates regularly.

## 6. Conclusion

The EDA identified important patterns in sales, profit, products, customers, cities, and order statuses. These findings help highlight areas where the business can investigate problems and develop improvement strategies.

The next steps are SQL analysis, Power BI dashboard development, and tracking business performance using measurable KPIs.

**Note:** These recommendations are based on the current dataset. Actual improvements in sales or profit must be measured after implementing business actions.
